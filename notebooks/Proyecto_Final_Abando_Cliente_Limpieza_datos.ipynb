## Modulo 1. Control de categorias, Limpieza de datos y Manejo de Valores Faltantes

Se identifican y corrigen errores y los valores nulos del dataset

```python
#La columna TotalCharges es de tipo texto cuando deberia ser del tipo numerico y esta tiene espacios vacios que impiden a isnull() detectar correctamente valores faltantes.

print('\n--------Trabajando con valores Nulos--------')
print(f'Total de espacios en blanco de la Columna TotalCharges: {(df['TotalCharges'] == ' ').sum()}')
print(f'Tipo de dato de la columna TotalCharges antes del cambio: {df['TotalCharges'].dtypes}')

#Conversion a tipo numerico con error='coerce', al encontrar un valor que no puede convertir a numero lo convertira en NaN
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
print(f'Tipo de dato de la columna TotalCharges Despues del cambio: {df['TotalCharges'].dtypes}')
```

```python
print('Total de valores faltantes encontrados: ')
nulos= df.isnull().sum()
porcentaje_nulos = ((df.isnull().sum() / len(df) * 100).round(2))

print("\nDatos Nulos:")
print(nulos)
print("\nPorcentaje de Datos Nulos")
```

## Limpieza de la columna TotalCharges

### Problema encontrado

La columna `TotalCharges` fue importada como tipo `object` debido a que contiene espacios en blanco en algunos registros. Estos espacios no son detectados por `isnull()` como valores faltantes.

### Decisión

Se convirtió la columna a tipo numérico utilizando `pd.to_numeric(errors='coerce')`, lo que permitió transformar los espacios vacíos en valores `NaN`.

Posteriormente se le explico al dueño de la empresa que eliminar esos registros es la mejor estrategia porque representan una pequeña cantidad de clientes y mantenerlos podría afectar el análisis.

```python
df = df.dropna(subset=['TotalCharges'])
print(f'Valores faltantes de la columna "TotalCharges": {df['TotalCharges'].isnull().sum()}')
print('Nueva forma del dataset', df.shape)
```

## Conversion de la columna Churn

### Problema encontrado

La columna mas importante del dataset, `Churn`,  fue importada como tipo `object` pero dicha variable deberia ser del tipo binario ya que solo muestra valores de "Si" y "No".

### Decisión

Se transformo la columna a una representacion binaria, asignandole el valor **1** a los clientes que abandonaron la empresa ("Yes") y el valor de **0** a los que permanecieron ("No").

### Justificacion

Los algoritmos de ML trabajan solo con datos numericos. Convertir la variable objectivo a formato binario ayuda a que el etrenamiento del modelo sea mas efectivo.

```python
df['Churn'] = df['Churn'].map({
    "No": 0,
    "Yes":1
}).astype('int64')
print(f'Nuevo tipo de dato para la columna "Churn": {df['Churn'].dtypes}')
print(df['Churn'].value_counts())
```

## Detección de Duplicados.

```python
print('\nBusqueda de valores duplicados')
print('Filas antes de verificar', len(df))
df = df.drop_duplicates()
print(f'Filas despues de verificar y eliminar valores duplicados, {len(df)}')
```

## Modulo 3. Feature Engineering

## Feature Engineering 1: Segmentación de la antigüedad del cliente

La variable `tenure` representa la cantidad de meses que un cliente ha permanecido en la empresa. Aunque esta información es útil, se decidió crear una nueva variable categórica denominada `TenureGroup` mediante una estrategia de discretización.

El objetivo es agrupar a los clientes según su nivel de antigüedad (Nuevo, Estable, Leal y Veterano), permitiendo que el modelo identifique patrones de abandono asociados a cada etapa del ciclo de vida del cliente.

```python
df['TenureGroup'] = pd.cut(
    df['tenure'],
    bins=[0,12,36,60,72],
    labels=('Nuevo', 'Estable', 'Leal', 'Veterano')
)


print(df['TenureGroup'].value_counts())
```

## Feature Engineering 2: Cantidad de servicios contratados

Los clientes pueden contratar múltiples servicios adicionales además del servicio principal de telefonía e Internet. En lugar de analizar cada servicio por separado, se creó la variable **TotalServices**, la cual representa la cantidad total de servicios contratados por cada cliente.

Esta nueva característica resume el nivel de utilización de los servicios de la empresa y permite evaluar si los clientes con un mayor número de servicios presentan una menor probabilidad de abandonar la compañía. Además, reduce la complejidad del análisis al concentrar varias variables relacionadas en un único indicador numérico.

```python
servicios = ['PhoneService', 'MultipleLines', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies' ]

for col in servicios:
    df[col] = df[col].replace({
        'Yes': 1,
        'No': 0,
        'No phone service': 0,
        'No internet service': 0
    })

df['TotalServices'] = df[servicios].sum(axis=1)

print(df[['TotalServices']].head())
print(f'Tipo de dato: {df['TotalServices'].dtypes}')
```
